# Court Case Analysis with DuckDB and PyGWalker

This notebook demonstrates how to query the court case data using DuckDB and visualize it interactively using PyGWalker.

```python
import duckdb
import pandas as pd
import pygwalker as pyg
```

## 1. Connect to DuckDB

We connect to the persistent DuckDB database where the data has been loaded.

```python
db_path = "/home/jovyan/court_cases.duckdb"
con = duckdb.connect(db_path)
```

## 2. Query Data

We can execute SQL queries directly against the DuckDB database and convert the result to a Pandas DataFrame.

```python
# Example Query: Select all records where the case was dismissed
query = """
SELECT 
    race, 
    gender, 
    attorneyType, 
    dispositionText, 
    netSentenceDays, 
    totalFinancial
FROM court_cases
LIMIT 10000
"""

df = con.execute(query).df()
print(f"Loaded {len(df)} rows.")
df.head()
```

## 3. Interactive Visualization with PyGWalker

PyGWalker turns your Pandas DataFrame into a Tableau-style user interface for visual exploration.

```python
walker = pyg.walk(df)
```

## 4. Close Connection

```python
con.close()
```
